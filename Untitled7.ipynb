```python
IMG_EXTS = (".jpg", ".jpeg", ".png")

def find_class_root(base):
    for root, dirs, files in os.walk(base):
        if dirs:
            sub = os.path.join(root, dirs[0])
            if os.path.isdir(sub) and any(f.lower().endswith(IMG_EXTS) for f in os.listdir(sub)):
                return root
    return None

data_dir = find_class_root("/content/data")
print("Data directory:", data_dir)

classes = sorted(d for d in os.listdir(data_dir) if os.path.isdir(os.path.join(data_dir, d)))
NUM_CLASSES = len(classes)
print(NUM_CLASSES, "classes:")
for c in classes:
    print("  ", c)
```

```python
paths, labels = [], []
for idx, c in enumerate(classes):
    folder = os.path.join(data_dir, c)
    for f in os.listdir(folder):
        if f.lower().endswith(IMG_EXTS):
            paths.append(os.path.join(folder, f))
            labels.append(idx)

paths = np.array(paths)
labels = np.array(labels)
print("Total images:", len(paths))
print("Example:", paths[0], "->", labels[0], classes[labels[0]])
```

```python
counts = np.bincount(labels, minlength=NUM_CLASSES)

plt.figure(figsize=(10, 6))
plt.barh(classes, counts, color="seagreen")
plt.xlabel("Number of images")
plt.title("Images per class")
plt.tight_layout()
plt.show()

for c, n in zip(classes, counts):
    print(f"{n:5d}  {c}")
```

```python
plt.figure(figsize=(14, 7))
for i, idx in enumerate(random.sample(range(NUM_CLASSES), 8)):
    img = Image.open(random.choice(paths[labels == idx]))
    plt.subplot(2, 4, i + 1)
    plt.imshow(img)
    plt.title(classes[idx], fontsize=8)
    plt.axis("off")
plt.tight_layout()
plt.show()
print("Image size example:", img.size)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_tmp, y_train, y_tmp = train_test_split(
    paths, labels, test_size=0.30, stratify=labels, random_state=SEED)
X_val, X_test, y_val, y_test = train_test_split(
    X_tmp, y_tmp, test_size=0.50, stratify=y_tmp, random_state=SEED)

print("Train:", len(X_train), "| Val:", len(X_val), "| Test:", len(X_test))
```

```python
IMG_SIZE = (224, 224)
BATCH = 32
AUTOTUNE = tf.data.AUTOTUNE

def load_image(path, label):
    img = tf.io.read_file(path)
    img = tf.io.decode_image(img, channels=3, expand_animations=False)
    img.set_shape([None, None, 3])
    img = tf.image.resize(img, IMG_SIZE)
    img = tf.cast(tf.round(img), tf.uint8)
    return img, label

def make_ds(X, y, training=False):
    ds = tf.data.Dataset.from_tensor_slices((X, y))
    ds = ds.map(load_image, num_parallel_calls=AUTOTUNE).cache()
    if training:
        ds = ds.shuffle(2000, seed=SEED)
    return ds.batch(BATCH).prefetch(AUTOTUNE)

train_ds = make_ds(X_train, y_train, training=True)
val_ds   = make_ds(X_val, y_val)
test_ds  = make_ds(X_test, y_test)
```

```python
for imgs, lbls in train_ds.take(1):
    print("Batch shape:", imgs.shape, "| dtype:", imgs.dtype)
    plt.figure(figsize=(14, 7))
    for i in range(8):
        plt.subplot(2, 4, i + 1)
        plt.imshow(imgs[i].numpy())
        plt.title(classes[int(lbls[i])], fontsize=8)
        plt.axis("off")
    plt.tight_layout()
    plt.show()
```

```python
from sklearn.utils.class_weight import compute_class_weight

cw = compute_class_weight(class_weight="balanced", classes=np.arange(NUM_CLASSES), y=y_train)
class_weight = dict(enumerate(cw))

for k, v in class_weight.items():
    print(f"{v:5.2f}  {classes[k]}")
```

```python
from tensorflow import keras
from tensorflow.keras import layers

data_aug = keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
], name="augmentation")
```

```python
def build_baseline():
    inputs = keras.Input(shape=(*IMG_SIZE, 3))
    x = layers.Rescaling(1.0 / 255)(inputs)
    x = data_aug(x)
    for filters in [32, 64, 128, 256]:
        x = layers.Conv2D(filters, 3, padding="same", use_bias=False)(x)
        x = layers.BatchNormalization()(x)
        x = layers.ReLU()(x)
        x = layers.MaxPooling2D()(x)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.4)(x)
    x = layers.Dense(128, activation="relu")(x)
    x = layers.Dropout(0.3)(x)
    outputs = layers.Dense(NUM_CLASSES, activation="softmax")(x)
    return keras.Model(inputs, outputs, name="baseline_cnn")

baseline = build_baseline()
baseline.compile(optimizer=keras.optimizers.Adam(1e-3),
                 loss="sparse_categorical_crossentropy",
                 metrics=["accuracy"])
baseline.summary()
```

```python
SAVE_DIR = "/content/drive/MyDrive/plant_disease_project"
os.makedirs(SAVE_DIR, exist_ok=True)

def get_callbacks(name):
    return [
        keras.callbacks.EarlyStopping(monitor="val_loss", patience=4, restore_best_weights=True),
        keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=2, min_lr=1e-6),
        keras.callbacks.ModelCheckpoint(f"{SAVE_DIR}/{name}_best.keras", monitor="val_loss", save_best_only=True),
    ]
```

```python
hist_base = baseline.fit(
    train_ds, validation_data=val_ds,
    epochs=20, class_weight=class_weight,
    callbacks=get_callbacks("baseline_cnn"))
```

```python
def plot_history(histories, title):
    acc, val_acc, loss, val_loss = [], [], [], []
    for h in histories:
        acc += h.history["accuracy"];  val_acc += h.history["val_accuracy"]
        loss += h.history["loss"];     val_loss += h.history["val_loss"]
    epochs = range(1, len(acc) + 1)
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    plt.plot(epochs, acc, label="train"); plt.plot(epochs, val_acc, label="validation")
    plt.title(f"{title}: accuracy"); plt.xlabel("epoch"); plt.legend()
    plt.subplot(1, 2, 2)
    plt.plot(epochs, loss, label="train"); plt.plot(epochs, val_loss, label="validation")
    plt.title(f"{title}: loss"); plt.xlabel("epoch"); plt.legend()
    plt.tight_layout(); plt.show()

plot_history([hist_base], "Baseline CNN")
```

```python
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix, f1_score

def evaluate_model(model, name):
    test_loss, test_acc = model.evaluate(test_ds, verbose=0)
    y_prob = model.predict(test_ds, verbose=0)
    y_pred = np.argmax(y_prob, axis=1)
    macro_f1 = f1_score(y_test, y_pred, average="macro")
    print(f"{name}: test accuracy = {test_acc:.4f} | macro F1 = {macro_f1:.4f}\n")
    print(classification_report(y_test, y_pred, target_names=classes, digits=3))

    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Greens",
                xticklabels=classes, yticklabels=classes, cbar=False)
    plt.xlabel("Predicted"); plt.ylabel("True")
    plt.title(f"Confusion matrix: {name}")
    plt.xticks(rotation=90); plt.tight_layout(); plt.show()
    return {"accuracy": test_acc, "macro_f1": macro_f1}

res_base = evaluate_model(baseline, "Baseline CNN")
```

```python
y_pred = np.argmax(baseline.predict(test_ds, verbose=0), axis=1)
cm = confusion_matrix(y_test, y_pred)
np.fill_diagonal(cm, 0)     # keep only the mistakes

pairs = [(cm[i, j], i, j) for i in range(NUM_CLASSES)
         for j in range(NUM_CLASSES) if cm[i, j] > 0]
for n, i, j in sorted(pairs, reverse=True)[:8]:
    print(f"{n:3d}  {classes[i]}  ->  predicted as  {classes[j]}")
```

```python
def build_mobilenet():
    base = keras.applications.MobileNetV2(
        input_shape=(*IMG_SIZE, 3), include_top=False, weights="imagenet")
    base.trainable = False                       # freeze the pretrained part

    inputs = keras.Input(shape=(*IMG_SIZE, 3))
    x = layers.Rescaling(1.0 / 127.5, offset=-1)(inputs)   # MobileNetV2 expects -1 to 1
    x = data_aug(x)
    x = base(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.3)(x)
    outputs = layers.Dense(NUM_CLASSES, activation="softmax")(x)
    return keras.Model(inputs, outputs, name="mobilenetv2"), base

mobilenet, base_model = build_mobilenet()
mobilenet.compile(optimizer=keras.optimizers.Adam(1e-3),
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])
mobilenet.summary()
```

```python
hist_head = mobilenet.fit(
    train_ds, validation_data=val_ds,
    epochs=8, class_weight=class_weight,
    callbacks=get_callbacks("mobilenet_head"))
```

```python
res_head = evaluate_model(mobilenet, "MobileNetV2 (head only)")
```

```python
base_model.trainable = True

for layer in base_model.layers[:-40]:        # keep everything except the top 40 frozen
    layer.trainable = False

for layer in base_model.layers:              # but never unfreeze BatchNorm layers
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False

mobilenet.compile(optimizer=keras.optimizers.Adam(1e-4),
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])

print("Trainable weights:", len(mobilenet.trainable_weights))
```

```python
rescale = [l for l in mobilenet.layers if isinstance(l, layers.Rescaling)][0]
gap     = [l for l in mobilenet.layers if isinstance(l, layers.GlobalAveragePooling2D)][0]
dense   = mobilenet.layers[-1]

def make_gradcam(img_array):
    x = tf.convert_to_tensor(img_array, dtype=tf.float32)       # shape (1, 224, 224, 3)
    with tf.GradientTape() as tape:
        tape.watch(x)
        conv_out = base_model(rescale(x), training=False)       # (1, 7, 7, 1280)
        preds = dense(gap(conv_out))                            # skip dropout and augmentation
        class_index = int(tf.argmax(preds[0]))
        class_score = preds[:, class_index]

    grads = tape.gradient(class_score, conv_out)
    weights = tf.reduce_mean(grads, axis=(1, 2))                # importance of each feature map
    cam = tf.reduce_sum(conv_out * weights[:, None, None, :], axis=-1)[0]
    cam = tf.nn.relu(cam)
    cam = cam / (tf.reduce_max(cam) + 1e-8)
    return cam.numpy(), preds.numpy()[0]

def show_gradcam(img_path):
    true_name = os.path.basename(os.path.dirname(img_path))     # folder name = true class
    img = Image.open(img_path).convert("RGB").resize(IMG_SIZE)
    cam, probs = make_gradcam(np.expand_dims(np.array(img, dtype=np.float32), 0))
    pred = int(np.argmax(probs))

    cam_big = np.array(Image.fromarray(np.uint8(cam * 255)).resize(IMG_SIZE, Image.BILINEAR)) / 255.0
    heat = plt.get_cmap("jet")(cam_big)[..., :3]
    overlay = 0.6 * np.array(img) / 255.0 + 0.4 * heat

    plt.figure(figsize=(10, 3.5))
    for i, (im, title) in enumerate([(img, "Original"), (cam_big, "Heatmap"), (overlay, "Overlay")]):
        plt.subplot(1, 3, i + 1)
        plt.imshow(im, cmap="jet" if title == "Heatmap" else None)
        plt.title(title); plt.axis("off")
    plt.suptitle(f"True: {true_name}\nPredicted: {classes[pred]} ({probs[pred]*100:.1f}%)", fontsize=9)
    plt.tight_layout(); plt.show()
```
